```python
weight = torch.tensor(1.0, requires_grad=True)

input_value = torch.tensor(2.0)
target = torch.tensor(10.0)

learning_rate = 0.1

for epoch in range(5):

    prediction = weight * input_value      #1×2=2

    loss = (prediction - target) ** 2      # (2 - 10) raise to power 2

    loss.backward() # backpropagation

    with torch.no_grad():
        weight -= learning_rate * weight.grad   # weight=1−(0.1×−32)

    weight.grad.zero_()

    print("Epoch:", epoch + 1)
    print("Prediction:", prediction.item())
    print("Loss:", loss.item())
    print("Weight:", weight.item())
    print()
```

```python
weight = torch.tensor(1.0, requires_grad=True)

input_value = torch.tensor(2.0)
target = torch.tensor(10.0)

optimizer = torch.optim.SGD([weight], lr=0.1)

for epoch in range(5):

    optimizer.zero_grad()

    prediction = weight * input_value

    loss = (prediction - target) ** 2

    loss.backward()

    optimizer.step()

    print("Epoch:", epoch + 1)
    print("Prediction:", prediction.item())
    print("Loss:", loss.item())
    print("Weight:", weight.item())
    print()
```

```python
linear_model = nn.Linear(
    in_features=1,
    out_features=1
)

print(linear_model)

print("Weight:", linear_model.weight)
print("Bias:", linear_model.bias)

print("Weight shape:", linear_model.weight.shape)
print("Bias shape:", linear_model.bias.shape)
```

```python
input_value = torch.tensor([[2.0]])

prediction = linear_model(input_value)

print("Input shape:", input_value.shape)
print("Prediction:", prediction)
print("Prediction shape:", prediction.shape)
```

```python
target = torch.tensor([[10.0]])

loss = (prediction - target) ** 2

print("Prediction", prediction)
print("Loss", loss)
print("Target", target)
```

```python
loss.backward()

print("Weight gradient:", linear_model.weight.grad)
print("Bias gradient:", linear_model.bias.grad)
```

```python
optimizer = torch.optim.SGD(
    linear_model.parameters(),
    lr=0.01
)

print(optimizer)
```

```python
optimizer.step()

print("Updated weight:", linear_model.weight)
print("Updated bias:", linear_model.bias)
```

```python
new_prediction = linear_model(input_value)
new_loss = (new_prediction - target) ** 2

print("Old prediction:", prediction)
print("New prediction:", new_prediction)
print("Old loss:", loss)
print("New loss:", new_loss)
```

```python
print("Weight gradient still stored:", linear_model.weight.grad)
print("Bias gradient still stored:", linear_model.bias.grad)
```

```python
optimizer.zero_grad()

print("Weight gradient after zero_grad:", linear_model.weight.grad)
print("Bias gradient after zero_grad:", linear_model.bias.grad)
```

```python
new_loss.backward()

print("New weight gradient:", linear_model.weight.grad)
print("New bias gradient:", linear_model.bias.grad)
```
