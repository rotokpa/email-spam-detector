```python
weight = torch.tensor(1.0, requires_grad=True)

input_value = torch.tensor(2.0)
target = torch.tensor(10.0)

learning_rate = 0.1

for epoch in range(5):

    prediction = weight * input_value      #1×2=2

    loss = (prediction - target) ** 2      # (2 - 10) raise to power 2

    loss.backward() # backpropagation

    with torch.no_grad():
        weight -= learning_rate * weight.grad   # weight=1−(0.1×−32)

    weight.grad.zero_()

    print("Epoch:", epoch + 1)
    print("Prediction:", prediction.item())
    print("Loss:", loss.item())
    print("Weight:", weight.item())
    print()
```

```python
from transformers import AutoModel

model = AutoModel.from_pretrained("distilbert-base-uncased") #loads the pretrained DistilBERT network and its learned weights
```

```python
print(type(model))
print("Model loaded:", model is not None)
```

```python
embedding_layer = model.embeddings #accesses the model's embedding module

print(type(embedding_layer))
```

```python
embedding_output = embedding_layer(encoded["input_ids"])

print("embedding shape:", embedding_output.shape)
```

```python
transformer_layers = model.transformer.layer

print("Number of Transformer layers:", len(transformer_layers))
print("First layer type:", type(transformer_layers[0]))
```

```python
first_layer = transformer_layers[0]
print(first_layer)
```

```python
attention = first_layer.attention

print(attention)
```

```python
attention = first_layer.attention

hidden_states = embedding_output

query = attention.q_lin(hidden_states)
key = attention.k_lin(hidden_states)
value = attention.v_lin(hidden_states)

print("Query shape:", query.shape)
print("Key shape:", key.shape)
print("Value shape:", value.shape)
```

```python
print("Hidden dimension:", model.config.dim)
print("Number of attention heads:", model.config.n_heads)

head_dimension = model.config.dim // model.config.n_heads

print("Dimension per head:", head_dimension)
```

```python
batch_size = query.shape[0]
sequence_length = query.shape[1]
number_of_heads = model.config.n_heads
head_dimension = model.config.dim // number_of_heads

query_heads = query.view(
    batch_size,
    sequence_length,
    number_of_heads,
    head_dimension
).transpose(1, 2)

key_heads = key.view(
    batch_size,
    sequence_length,
    number_of_heads,
    head_dimension
).transpose(1, 2)

print("Query heads shape:", query_heads.shape)
print("Key heads shape:", key_heads.shape)
```
