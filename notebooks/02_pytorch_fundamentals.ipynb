```python
x = torch.tensor(3.0, requires_grad=True)

y = x ** 2

print(x)
print(y)
```

```python
y.backward()

print(x.grad)
```

```python
x1 = 3.0
x2 = 3.001

y1 = x1 ** 2
y2 = x2 ** 2

change_in_x = x2 - x1
change_in_y = y2 - y1

estimated_gradient = change_in_y / change_in_x

print(change_in_x)
print(change_in_y)
print(estimated_gradient)
```

```python
weight = torch.tensor(1.0, requires_grad=True)

input_value = torch.tensor(2.0)

target = torch.tensor(10.0)

prediction = weight * input_value

loss = (prediction - target) ** 2

print("Prediction:", prediction)
print("Loss:", loss)
```

```python
loss.backward()

print(weight.grad)
```

```python
learning_rate = 0.1

with torch.no_grad():
    weight -= learning_rate * weight.grad
    
print("Updated weight:", weight)
```

```python
prediction = weight * input_value 

loss = (prediction - target) ** 2

print("Prediction:", prediction)
print("Loss:", loss)
```

```python
weight.grad.zero_()

 # reset gradient

print(weight.grad)
```

```python
loss.backward()

print(weight.grad)
```

```python
with torch.no_grad():
    weight -= learning_rate * weight.grad # update new weight 

print("Updated weight:", weight)
```

```python
prediction = weight * input_value

loss = (prediction - target) ** 2

print("Prediction:", prediction)
print("Loss:", loss)
```
