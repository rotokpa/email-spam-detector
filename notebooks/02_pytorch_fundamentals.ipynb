```python
weight = torch.tensor(1.0, requires_grad=True)

input_value = torch.tensor(2.0)
target = torch.tensor(10.0)

learning_rate = 0.1

for epoch in range(5):

    prediction = weight * input_value      #1×2=2

    loss = (prediction - target) ** 2      # (2 - 10) raise to power 2

    loss.backward() # backpropagation

    with torch.no_grad():
        weight -= learning_rate * weight.grad   # weight=1−(0.1×−32)

    weight.grad.zero_()

    print("Epoch:", epoch + 1)
    print("Prediction:", prediction.item())
    print("Loss:", loss.item())
    print("Weight:", weight.item())
    print()
```

```python
inputs = torch.tensor([
    [1.0],
    [2.0],
    [3.0],
    [4.0]
])

targets = torch.tensor([
    [5.0],
    [10.0],
    [15.0],
    [20.0]
])

print("Inputs:")
print(inputs)

print("Input shape:", inputs.shape)

print("\nTargets:")
print(targets)

print("Target shape:", targets.shape)
```

```python
torch.manual_seed(42)

multi_model = nn.Linear(
    in_features=1,
    out_features=1
)

predictions = multi_model(inputs)

print("Starting weight:", multi_model.weight)
print("Starting bias:", multi_model.bias)

print("\nPredictions:")
print(predictions)

print("\nPrediction shape:", predictions.shape)
```

```python
squared_errors = (predictions - targets) ** 2   # calculating losses

print("Squared errors:")
print(squared_errors)

print("Squared errors shape:", squared_errors.shape)
```

```python
loss = squared_errors.mean()  # loss average sum of value by numbers of value

print("Mean squared error:", loss)
print("Loss shape:", loss.shape)
print("Loss dimensions:", loss.ndim)
```

```python
mse_loss = nn.MSELoss()

loss_from_pytorch = mse_loss(predictions, targets)

print("Manual MSE:", loss)
print("PyTorch MSE:", loss_from_pytorch)
```

```python
print("Weight gradient before backward:", multi_model.weight.grad)
print("Bias gradient before backward:", multi_model.bias.grad)
```

```python
loss_from_pytorch.backward() # propagation

print("Weight gradient:", multi_model.weight.grad)
print("Bias gradient:", multi_model.bias.grad)
```

```python
optimizer.step()

print("Updated weight:", multi_model.weight)
print("Updated bias:", multi_model.bias)
```

```python
torch.manual_seed(42)

multi_model = nn.Linear(
    in_features=1,
    out_features=1
)

mse_loss = nn.MSELoss()

optimizer = torch.optim.SGD(
    multi_model.parameters(),
    lr=0.01
)
```

```python
for epoch in range(100):
    epoch_loss = 0.0
    
    for batch_inputs, batch_targets in dataloader:
        optimizer.zero_grad()
        prediction = multi_model(batch_inputs)
        loss = mse_loss(prediction, batch_targets)
        epoch_loss += loss.item()
        loss.backward()
        optimizer.step()
    average_loss = epoch_loss / len(dataloader)
```
